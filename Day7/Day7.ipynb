```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
import chromadb

print("All libraries imported successfully!")
print(f'ChromaDB version: {chromadb.__version__}')
```

```python
documents=[
    "ETL is used to clean and transform data",  #doc_0
    "A vechile is a mode of transportation",  #doc_1
    "Cars and trucks are popular automobile",  #doc_2
    "SQL is used to query database",    #doc_3
    "Machine learning trains models on data"   #doc_4
]

query_keyword="vechile"

print("="*60)
print(f"KEYWORD SEARCH for:'{query_keyword}'")
print("="*60)

for i,doc in enumerate(documents):
  if query_keyword.lower() in doc.lower():
    print(f" FOUND  [doc_{i}]:{doc}")
  else:
    print(f" MISSED [doc_{i}]:{doc}")
print()
print("PROBLEM:doc_2 talks about 'Cars and trucks'-which ARE vechicles!")
print("But keywords search MISSED it because it searcher for the exact word'vechile'.")
```

```python
print("Loading embedding model...(may take 1-2 mins on 1st run)")
model=SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded successfully!")
print(f"Model produces vectory of size:{model.get_sentence_embedding_dimension()} dimensions")
```

```python
sentence="ETL is used to clean and transform data"

embedding=model.encode(sentence)
print(f"Input sentence:'{sentence}'")
print()
print(f"Embedding type:{type(embedding)}")
print(f"Embedding shape:{embedding.shape}")
print(f"First 10 numbers:{embedding[:10].round(4)}")
print(f"Min valuue:{embedding.min():.4f}")
print(f"Max valuue:{embedding.max():.4f}")
```

```python
sentences = [
    "ETL is used to clean and transfrom data",    #doc_1
    "A vehicle is a mode of transport",            #doc_2
    "cars and trucks are popular transport",      #doc_3
    "SQL is used to query databases",             #doc_4
    "Machine learning trains model on data"       #doc_5

]

embedding = model.encode(sentences)

print(f'input sentence: {sentences}')
print()
print(f'embedding format {type(embedding)}')

print(f'embedding shape {embedding.shape}')

print(f"First 10 numbers:{embedding[:10].round(4)}")
print(f"Min valuue:{embedding.min():.4f}")
print(f"Max valuue:{embedding.max():.4f}")

query_keyword='vehicle'
query_embedding = model.encode([query_keyword])
similarities = np.dot(query_embedding, embedding.T)
similarities = similarities[0]
similarity_threshold = 0.5

print("\nSemantic Search Results:")
for i , doc in enumerate(sentences):
  current_similarity = similarities[i]
  if current_similarity >= similarity_threshold:
    print(f'found  [doc_{i}] (Similarity: {current_similarity:.4f}): {doc}')
  else:
    print(f'missed [doc_{i}] (Similarity: {current_similarity:.4f}): {doc}')

print()
```

```python
your_sentence = [
    "I like travel the world",
    "I want to go different country",
    "I want vacation this year"
    "Enjoy the life"
]
key_words = ["vacation"]

your_sentence_embedding = model.encode(your_sentence)
key_words_embedding = model.encode(key_words)

similarity_matrix = np.dot(key_words_embedding, your_sentence_embedding.T)
similarity_threshold = 0.45
print("Keyword Semantic Search Results")
for i, keyword in enumerate(key_words):
    print(f"Keyword: '{keyword}'")
    print("-" * 30)
    for j, sentence in enumerate(your_sentence):
        similarity_score = similarity_matrix[i][j]
        if similarity_score >= similarity_threshold:
            print(f"  FOUND  | Score: {similarity_score:.4f} | '{sentence}'")
        else:
            print(f"  MISSED | Score: {similarity_score:.4f} | '{sentence}'")
    print()
```

```python
chroma_client=chromadb.Client()

collection=chroma_client.get_or_create_collection("demo_notes")

print("ChromaDB client created(in-memory model)")
print(f"Collection name:demo_name")
print(f"Documents in collection:{collection.count()}")
```

```python
sample_docs=[
    "ETL is used to clean and transfrom data",    #doc_1
    "A vehicle is a mode of transport",            #doc_2
    "cars and trucks are popular transport",      #doc_3
    "SQL is used to query databases",             #doc_4
    "Machine learning trains model on data"       #doc_5

]

sample_ids=["doc_1","doc_2","doc_3","doc_4","doc_5"]
#sample_ids=["doc001","doc002","doc003","doc004","doc005"]

sample_metadata=[
    {"subject":"Data Engineering","topic":"ETL"},
    {"subject":"Data Engineering","topic":"SQL"},
    {"subject":"ML","topic":"ML Bacis"},
    {"subject":"Python","topic":"Pandas"},
    {"subject":"ML","topic":"Neural Networks"}
]
collection.add(
    documents=sample_docs,
    ids=sample_ids,
    metadatas=sample_metadata
)

print(f"Documents adder to collection!")
print(f"Total doc now in collection:{collection.count()}")
```

```python
quer="How do I clean and prepare data?"

results=collection.query(
    query_texts=[quer],
    n_results=3
)

print("RESULT KEYS AVAILABLE:")
print(list(results.keys()))
```

```python
print(f"Query:'{quer}'")
print("="*60)
print()
matched_docs=results['documents'][0]
matched_ids=results['ids'][0]
matched_distances=results['distances'][0]
matched_metadata=results['metadatas'][0]

for rank,(doc,doc_id,dist,meta) in enumerate(zip(matched_docs,matched_ids,matched_distances,matched_metadata),start=1):
  print(f"Rank{rank}|ID:{doc_id}|Distance:{dist:.4f}")
  print(f"Subject:{meta['subject']}|Topic:{meta['topic']}")
  print(f"Document:{doc}")
  print()
  print("NOTICE:The result are above ETL & pandas")
  print("Semantic search found them even though the words are different")
```
